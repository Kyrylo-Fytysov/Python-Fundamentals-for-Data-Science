```python
def is_prime(num):
    "Функція для перевірки, чи є число простим."
    if num < 2:
        return False
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            return False
    return True

def start_task():
    "Функція для знаходження простих чисел на відрізку."
    user_input = input("Введіть 2 числа a і b: ")
    parts = user_input.split()

    if len(parts) != 2:
        print("Помилка: Потрібно ввести два числа через пробіл.")
        return

    try:
        a = int(parts[0])
        b = int(parts[1])
    except ValueError:
        print("Помилка: У списку присутнє нечислове значення")
        return

    start, end = min(a, b), max(a, b)
    primes = [n for n in range(start, end + 1) if is_prime(n)]
    print(primes)
```

```python
start_task()
```
